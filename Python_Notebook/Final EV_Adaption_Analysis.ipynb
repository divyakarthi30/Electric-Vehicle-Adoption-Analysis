# **Stage 2 – Data Cleaning and Pre-processing**

## **🎯 Objective**

To prepare the dataset by ensuring data quality, consistency, and usability for accurate analysis.

**🔍 Step 1 – Handling Missing Values**


The dataset was checked for missing values. No null values were found, so no imputation or removal was required.

```python
# Check missing values
df.isnull().sum()
```

**🔁 Step 2 – Handling Duplicate Records**

Duplicate records were checked and removed to ensure data accuracy and avoid redundancy.

```python
# Check duplicates
df.duplicated().sum()
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)
```

**🔄 Step 3 – Data Type Verification and Conversion**


Data types verification:  Data types were verified.

Data Convertion        :  Categorical data (Yes/No) was converted into numerical format for analysis.

```python
# Check data types
df.dtypes
```

**📊 Step 4 – Outlier Detection**



```python
import matplotlib.pyplot as plt

# Boxplot for outlier detection
df.boxplot(figsize=(25,7))
plt.title("Outlier Detection")
plt.show()
```

Outliers were identified using boxplots. Since the data is real-world, extreme values were retained.

📈 **Step 5 – Skewness Check**

```python
df['govt_incentives'] = df['govt_incentives'].astype(str).str.strip().str.lower()
```

```python
df['govt_incentives'] = df['govt_incentives'].map({'yes': 1, 'no': 0})
```

```python
# Check skewness
df.skew(numeric_only=True)
```

Skewness was checked to understand data distribution. No major skewness was found, so no transformation was applied.

⚙️Step 6 – Feature Engineering
 Explanation


 New features were created to improve analysis:

EV per charging station → identifies infrastructure gap


Cost per KM → evaluates cost efficiency

```python
# Create new features

# EV per Charging Station
df['EV_per_Station'] = df['ev_sales_units'] / df['charging_stations']
```

```python
# Cost per KM
df['Cost_per_KM'] = df['avg_ev_price_usd'] / df['vehicle_range_km']
```

# **🔹 Stage 3 – EDA and Visualizations**
### **Purpose of EDA**


*   To understand data patterns and trends

*   To identify relationships between variables

*   To detect anomalies and outliers

*   To support data-driven decision making



### **1. Univariate Analysis (Single Variable)**

* Analyze distribution of individual variables
     

* Use charts like histogram, boxplot

* Helps understand spread, frequency, and outliers


* Example: EV sales distribution, price variation


**1. Histogram For Numeric Distribution**

   

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['ev_sales_units'], bins=20)
plt.title("Distribution of EV Sales")
plt.xlabel("EV Sales")
plt.ylabel("Frequency")
plt.show()
```

## Interpretation

This histogram shows the distribution of EV sales.

Most values are concentrated in a certain range, indicating that EV sales are not evenly distributed.

Feature used: ev_sales_units

Insight: EV sales vary across regions, showing uneven adoption.


 **2.Boxplot for Outliers**

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['avg_ev_price_usd'])
plt.title("Boxplot of EV Price")
plt.show()
```

## Interpretation

This boxplot shows the spread of EV prices and identifies outliers.

Feature used: avg_ev_price_usd

Insight: Some EVs are significantly more expensive, indicating premium market segments.

## **2. Bivariate Analysis (Two Variables)**
     

*   Study relationship between two variables


*   Use scatter plots and bar charts


*   Helps identify correlation and trends

*   Example: Charging stations vs EV sales


### 1.Scatter Plot

```python
plt.figure(figsize=(8,5))
plt.scatter(df['charging_stations'], df['ev_sales_units'])
plt.xlabel("Charging Stations")
plt.ylabel("EV Sales")
plt.title("Charging Stations vs EV Sales")
plt.show()
```

### Interpretation

This chart shows the relationship between charging infrastructure and EV sales.

Features used: charging_stations, ev_sales_units

Insight: As charging stations increase, EV sales also increase, showing infrastructure impact.


**2.Bar Plot**

```python
df.groupby('govt_incentives')['ev_sales_units'].mean().plot(kind='bar')
plt.title("Impact of Government Incentives on EV Sales")
plt.xlabel("Government Incentives")
plt.ylabel("Average EV Sales")
plt.show()
```

### Interpretation

This bar chart compares EV sales based on government incentives.

Features used: govt_incentives, ev_sales_units

Insight: Regions with incentives have higher EV adoption

## **3. Multivariate Analysis- Correlation Heatmap(Multiple Variables)**

* Analyze relationships among multiple variables
* Use heatmaps, pairplots
* Helps understand complex patterns
* Example: Impact of price, range, and infrastructure together


**1.Heatmaps**

```python
import seaborn as sns

plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

This heatmap shows relationships between multiple variables.

Features used: all numeric columns

Insight: Strong positive correlation between charging stations and EV sales indicates infrastructure importance.

**2.Pair Plot**

```python
import seaborn as sns
import matplotlib.pyplot as plt

cols = ['ev_sales_units', 'charging_stations', 'avg_ev_price_usd', 'vehicle_range_km']

sns.pairplot(df[cols])
plt.show()
```

## Interpretation

* This pair plot shows the relationships between multiple numerical variables in the dataset using scatter plots and distributions.

* It helps in understanding how each feature is related to others.

* Features used: ev_sales_units, charging_stations, avg_ev_price_usd, vehicle_range_km

* Insight: There is a positive relationship between charging stations and EV sales, indicating that better infrastructure leads to higher adoption.

* Other variables such as price and vehicle range show moderate relationships, suggesting they also influence EV adoption to some extent.
